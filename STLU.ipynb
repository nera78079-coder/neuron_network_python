```python
import numpy as np
```

```python
class TLU:
    
    w=.5

    def __init__(self):
        pass 

    def sigmoid (self,x):
        return 1/(1+np.exp(-x))

    def tanh (self,x):
        return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

    def caculate_z_funtion (self,x):
        return x*self.w

    def caculate_activation_function (self,x,act_n):
        if act_n=="sigmoid":
            return self.sigmoid(self.caculate_z_funtion(x))
        else:
            return self.tanh(self.caculate_z_funtion(x))
    
```

```python
tlu=TLU()
x=10
act="sigmoid"
print("z :",tlu.caculate_z_funtion(x))
print("sigmoid :",tlu.caculate_activation_function(x,act))
act="tanh"
print("tanh :",tlu.caculate_activation_function(x,act))
```
